### Initial Observations

**Relationship between `date`, `d`, and `wm_yr_wk`:**

> `date` is simply the calendar date, starting from 2011-01-29. `d`, is the count of days starting from 2011-01-29 (starting at 1). `wm_yr_wk` is short for "WalMart Year Week". It is formatted xYYWW, where x is always 1 for a currently unknown reason. YY is their retail year, which extends into the following calendar year's January. Then WW is their retail week which typically goes from 01-52, but some years will go from 01-53, depending on where the weekdays land (the cutoff for starting the new year might be whether 1/26 is a Saturday, based on the 2013 and 2014 date encoding patterns).

**Why do all three exist?**

> They all represent different but independently useful ways to represent dates and timeframes.

**Role of `event_name_*` and `event_type_*`:**

> These represent significant holidays/events that may affect consumer activity and sales. The holidays/events are categorized as one of "Sporting", "Cultural", "National", or "Religious".

**Role of `snap_CA`, `snap_TX`, and `snap_WI`:**

> At this time, the exact role of these fields is unclear to me. I can see that they are binary/boolean representations of some common feature for each of the 3 states represented (California, Texas, and Wisconsin), though I don't know exactly what "snap" means.

**Anything else that stands out?**

```python
price_change_df = pd.DataFrame(columns=["store_id", "item_id", "number_of_price_changes"])
for store in sell_prices["store_id"].unique():
    for item in sell_prices["item_id"].unique():
        single_product_test_df = sell_prices[
            ((sell_prices["store_id"] == store) 
            & (sell_prices["item_id"] == item))
        ]
        number_of_price_changes = len(
                single_product_test_df[
                single_product_test_df["sell_price"] 
                != single_product_test_df["sell_price"].shift(
                    fill_value=single_product_test_df["sell_price"].iloc[0]
                )
            ]
        )
        price_change_df.loc[len(price_change_df)] = [store, item, number_of_price_changes]
```

```python
groups = sell_prices.groupby(["store_id", "item_id"])
```

```python
for (store, item), group in groups:
    print(store)
    print(item)
    print(group.head())
    print(type(group))
    break
```

```python
data_list = []
for (store, item), group in groups:
    number_of_price_changes = len(
        group[
            group["sell_price"] 
            != group["sell_price"].shift(
                fill_value=group["sell_price"].iloc[0]
            )
        ]
    )
    data_list.append([store, item, number_of_price_changes])
```

```python
price_change_df = pd.DataFrame(
    columns=["store_id", "item_id", "number_of_price_changes"],
    data=data_list
)
```
